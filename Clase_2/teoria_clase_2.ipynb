## Bucles

> For, es una estructura de iteración, recorre todo el elemento 


```python
learners = [
    {"nombre": "Pepita", "promedio": 8.7},
    {"nombre": "Pepito", "promedio": 7.5},
    {"nombre": "Andres", "promedio": 2},
    {"nombre": "Juan", "promedio": 8.7},
    {"nombre": "Cesar", "promedio": 3.5},
    {"nombre": "Carlos", "promedio": 3.5},
    {"nombre": "Cami", "promedio": 2},
    {"nombre": "Pepito", "promedio": 7.5},
    {"nombre": "Andres", "promedio": 9.1},
    {"nombre": "Pepita", "promedio": 2},
    {"nombre": "Pepito", "promedio": 7.5},
    {"nombre": "Andres", "promedio": 3.5},
    {"nombre": "Pepita", "promedio": 8.7},
    {"nombre": "Pepito", "promedio": 7.5},
    {"nombre": "Andres", "promedio": 5},
]

for learner in learners:
    if learner["promedio"] >= 7:
        print(f"Felicitaiones {learner['nombre']}")
    elif learner["promedio"] >=5:
        print(f"{learner['nombre']} Puedes mejorar")
    else:
        print(f"Necesitas mejorar")
```

```python
def evaluar_estudiante(nombre, promedio=5.0):
    if promedio >= 7:
        return f"{nombre}: ¡Excelente!"
    elif promedio >= 5:
        return f"{nombre}: Sigue así."
    else:
        return f"{nombre}: Refuerza los temas."
```

```python
print(evaluar_estudiante("Laura", 8.7))
print(evaluar_estudiante("Carlos", 5.2))
print(evaluar_estudiante("Anita"))  
```

```python
estudiantes = [
    {"nombre": "Laura",  "promedio": 8.7},
    {"nombre": "Carlos", "promedio": 5.2},
    {"nombre": "Anita",  "promedio": 9.1}
]

for estudiante in estudiantes:
    resultado = evaluar_estudiante(estudiante["nombre"], estudiante["promedio"])
    print(resultado)
```

```python
def registrar_estudiante(nombre, edad, correo):
    return {
        "nombre": nombre,
        "edad":   edad,
        "correo": correo,
        "activo": True
    }

nuevo = registrar_estudiante("Ana", 25, "ana@mail.com")
nuevo2 = registrar_estudiante("Pedro", 25, "pedro@mail.com")
print(nuevo)
print(nuevo2)
```

```python
import pandas as pd
curso_ia = [
    registrar_estudiante("Laura",  28, "laura@mail.com"),
    registrar_estudiante("Carlos", 32, "carlos@mail.com"),
    registrar_estudiante("Anita",  25, "anita@mail.com"),
]

curso_ia[0]["promedio"] = 8.7
curso_ia[1]["promedio"] = 5.2
curso_ia[2]["promedio"] = 9.1
```

```python
df = pd.DataFrame(curso_ia)
print(df)
```

```python
print("Promedio general:", df["promedio"].mean())
print("Mejor promedio:  ", df["promedio"].max())
print("Estudiante destacado:", df.loc[df["promedio"].idxmax(), "nombre"])
```
